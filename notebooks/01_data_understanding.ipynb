```python
# ==========================================
# INTELLIGENT IT OPERATIONS
# STEP 1: DATA UNDERSTANDING
# ==========================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')

print("Libraries imported successfully!")
```

```python
# Define dataset paths

issues_path = "../data/raw/issues.csv"

utterances_path = "../data/raw/sample_utterances.csv"

history_path = "../data/raw/issues_change_history.csv"

snapshot_path = "../data/raw/issues_snapshot.csv"

# Load datasets

issues = pd.read_csv(issues_path)

utterances = pd.read_csv(utterances_path)

history = pd.read_csv(history_path)

snapshot = pd.read_csv(snapshot_path)

print("All datasets loaded successfully!")
```

```python
#Check dataset dimensions
print("ISSUES DATASET:", issues.shape)

print("UTTERANCES DATASET:", utterances.shape)

print("CHANGE HISTORY DATASET:", history.shape)

print("SNAPSHOT DATASET:", snapshot.shape)
```

```python
#Inspect column names
print("ISSUES COLUMNS:")
print(issues.columns.tolist())

print("\nUTTERANCES COLUMNS:")
print(utterances.columns.tolist())

print("\nHISTORY COLUMNS:")
print(history.columns.tolist())

print("\nSNAPSHOT COLUMNS:")
print(snapshot.columns.tolist())
```

```python
#Check data types
print("ISSUES DATA TYPES")
print(issues.dtypes)

print("\nUTTERANCES DATA TYPES")
print(utterances.dtypes)

print("\nHISTORY DATA TYPES")
print(history.dtypes)

print("\nSNAPSHOT DATA TYPES")
print(snapshot.dtypes)
```

```python
#Check missing values
datasets = {
    "Issues": issues,
    "Utterances": utterances,
    "History": history,
    "Snapshot": snapshot
}

for name, df in datasets.items():

    print(f"\n========== {name} ==========")

    missing = df.isnull().sum()

    missing = missing[missing > 0]

    print(missing.sort_values(ascending=False))
```

```python
#Check duplicate records
for name, df in datasets.items():
    

    print(f"{name}: {df.duplicated().sum()} duplicate rows")
```

```python
#Check unique issue IDs
print("Unique issue IDs:", issues['id'].nunique())

print("Unique utterance issue IDs:", utterances['issueid'].nunique())

print("Unique history issue IDs:", history['issueid'].nunique())

print("Unique snapshot issue IDs:", snapshot['id'].nunique())
```
